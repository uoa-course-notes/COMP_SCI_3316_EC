```python
print(f"=========== Running experiments for algorithm: {gsemo.name} ========== ")
# start time - use perf_counter for wall-clock time
start_time = time.perf_counter()
# create a new experiment for the current algorithm 
experiment = ioh.Experiment(
    algorithm=gsemo,
    algorithm_name=gsemo.name,
    algorithm_info=gsemo.algorithm_info,
    fids = config.PROBLEM_IDS,
    iids = [1], 
    dims=[config.DIMENSION],
    reps=config.REPETITIONS,
    problem_class=config.PROBLEMS_TYPE,  # Use the configured problem class # type: ignore
    old_logger=False,  # type: ignore
    output_directory=str(out_base),
    folder_name=f"ioh-data-{gsemo.name}",
    zip_output=True, 
)

# run the experiment
experiment.run()

end_time = time.perf_counter()
elapsed_time = end_time - start_time
elapsed_times.append(elapsed_time)
# convert to minutes
elapsed_time_minutes = elapsed_time / 60
print(f"Time taken for {gsemo.name}: {elapsed_time:.2f} seconds ({elapsed_time_minutes:.2f} minutes)")
```
